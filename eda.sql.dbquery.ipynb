```sql
WITH no_show_counts AS (
SELECT
  PatientId,
  SUM(
    CASE
      WHEN Showed_up = 'false' THEN 1
      ELSE 0
    END
  ) AS no_show_count
FROM
  workspace.default.bronze_features
WHERE
  Showed_up != 'true'
GROUP BY
  PatientId
ORDER BY no_show_count desc)

SELECT COUNT(*)
FROM no_show_counts
WHERE no_show_count = 5;


-- 1: 13934
--2: 2356
-- 3: 512
-- 4: 162
-- 5: 58


```
